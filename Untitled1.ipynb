```python
import pandas as pd
import numpy as np

# Set random seed for reproducibility
np.random.seed(42)
n_rows = 1500

# Generate realistic data
cities = ['Rome', 'Milan', 'Florence']
property_types = ['Entire Home', 'Private Room', 'Shared Room']

data = {
    'Listing_ID': range(10001, 10001 + n_rows),
    'City': np.random.choice(cities, n_rows, p=[0.5, 0.3, 0.2]),
    'Room_Type': np.random.choice(property_types, n_rows, p=[0.7, 0.25, 0.05]),
    'Price_Per_Night_Euros': np.random.randint(40, 350, n_rows),
    'Availability_365': np.random.randint(10, 365, n_rows),
    'Minimum_Nights': np.random.choice([1, 2, 3, 5, 7], n_rows, p=[0.3, 0.4, 0.15, 0.1, 0.05]),
    'Number_of_Reviews': np.random.randint(0, 150, n_rows),
    'Host_Total_Listings': np.random.choice([1, 2, 3, 7, 12], n_rows, p=[0.6, 0.2, 0.1, 0.07, 0.03])
}

df = pd.DataFrame(data)

# Inject an economic trend: Entire homes available year-round have higher prices
df.loc[(df['Room_Type'] == 'Entire Home') & (df['Availability_365'] > 200), 'Price_Per_Night_Euros'] += 50

# Save to CSV
df.to_csv('italian_housing_data.csv', index=False)
print("🎯 Success! 'italian_housing_data.csv' has been created in your workspace directory.")
```

```python
# 1. Average price per night by city
# Goal: To discover where tourists (and hosts) are spending the most money (i.e the most expensive city and its average rental yield)
print("--- Question 1: Avg Price by City ---")
print(df.groupby('City')['Price_Per_Night_Euros'].mean().round(2))

# 2. Percentage of commercial multi-hosts (hosts with > 5 listings)
# Goal: Prove whether big commercial players are buying up housing stock instead of local residents (i.e are commercial multi-hosts dominating the market?)
print("\n--- Question 2: Commercial Multi-Hosts ---")
multi_hosts = df[df['Host_Total_Listings'] > 5].shape[0]
pct_multi = (multi_hosts / df.shape[0]) * 100
print(f"{pct_multi:.2f}% of listings are controlled by commercial multi-hosts")

# 3. Find affordable private rooms for long-term residents
# Goal: Identify which city or property types are least affected by high tourist pricing (i.e where should a local resident look for long-term housing?
print("\n--- Question 3: Top 5 Most Affordable Private Rooms ---")
private_rooms = df[df['Room_Type'] == 'Private Room'].sort_values(by='Price_Per_Night_Euros')
print(private_rooms[['City', 'Price_Per_Night_Euros', 'Availability_365']].head(5))
```

```python
# Interpretations from our result: 
# 1. Low Baseline Prices: The most affordable private rooms start at €41 to €44 per night in Florence and Rome.
# 2. The Availability Trap (Row 59): Notice how listing 59 in Rome is available 238 days out of the year. Even though it is cheap, it is being hoarded for tourists for nearly 8 months out of the year, reducing long-term housing options for local students or workers.
# 3. Duplicate Trends (Rows 1230 & 1052): Both are in Florence, priced at €43, with 178 days of availability. This indicates they are likely managed by the exact same commercial host operating identical units.
```
